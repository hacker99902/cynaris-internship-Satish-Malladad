# W1D3: Data Loading, Cleaning and Inspection

## Objective
To load, clean and inspect a COVID-19 dataset using Pandas and NumPy.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/states.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
df.info()
```

```python
# total null val in dataset
df.isnull().sum()
```

```python
# check for duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.dropna(how="all")
df = df.dropna(axis=1, how="all")
```

```python
# rm extra spaces 
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.strip()

df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df.head()
```

```python
# Handle missing numeric values
numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nStatistical summary:")
df.describe()
```

```python
df.to_csv("../data/cleaned/states_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
