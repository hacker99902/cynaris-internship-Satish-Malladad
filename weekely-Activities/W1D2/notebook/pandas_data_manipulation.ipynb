# W1D2 – Pandas Data Manipulation


## 1. Import Pandas

```python
import pandas as pd
import os
```

## 2. Load the dataset

```python
file_path = "data/states.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
df.head()
```

## 3. Explore the dataset

```python
print("Shape (rows, columns):", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types and non-null counts:")
df.info()
```

```python
print("Summary statistics:")
display(df.describe(include="all"))

print("Missing values in each column:")
df.isnull().sum()
```

## 4. Clean the dataset

```python
# Make a copy so the original DataFrame remains unchanged
clean_df = df.copy()

# Remove completely duplicated rows
clean_df = clean_df.drop_duplicates()

# Fill missing numeric values with the column median
numeric_cols = clean_df.select_dtypes(include="number").columns
for column in numeric_cols:
    clean_df[column] = clean_df[column].fillna(clean_df[column].median())

# Fill missing text values with "Unknown"
text_cols = clean_df.select_dtypes(include=["object", "string"]).columns
for column in text_cols:
    clean_df[column] = clean_df[column].fillna("Unknown")

print("Original shape:", df.shape)
print("Cleaned shape:", clean_df.shape)
clean_df.head()
```

## 5. Select and filter data

```python
# Select the first two columns (works even if the dataset has fewer columns)
clean_df.iloc[:, :2].head()
```

```python
# Example: filter rows using the first numeric column, if one exists
if len(numeric_cols) > 0:
    column = numeric_cols[0]
    threshold = clean_df[column].median()
    filtered_df = clean_df[clean_df[column] >= threshold]
    print(f"Rows where {column} >= {threshold}:")
    display(filtered_df.head())
else:
    print("No numeric column is available for this filter example.")
```

## 6. Sort and group data

```python
if len(numeric_cols) > 0:
    column = numeric_cols[0]
    display(clean_df.sort_values(by=column, ascending=False).head())
else:
    print("No numeric column is available for sorting.")
```

```python
# Group by the first text column and count rows in each group
if len(text_cols) > 0:
    group_column = text_cols[0]
    display(clean_df.groupby(group_column).size().reset_index(name="count").head(10))
else:
    print("No text column is available for this grouping example.")
```

## 7. Save the cleaned dataset

```python
os.makedirs("output", exist_ok=True)
output_path = "output/cleaned_states.csv"
clean_df.to_csv(output_path, index=False)
print("Cleaned dataset saved to:", output_path)
```
